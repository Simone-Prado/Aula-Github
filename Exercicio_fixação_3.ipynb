```python
#1. Exercício 1 – Média da turma Crie um array NumPy com as notas: [6.5, 8.0, 7.5, 9.0, 5.0, 8.5] Calcule a média da turma e utilize if/else para exibir:
#• "Turma com bom desempenho" se a média for maior ou igual a 7;
#• "Turma precisa melhorar" caso contrário.

import numpy as np

notas = np.array([6.5, 8.0, 7.5, 9.0, 5.0, 8.5])
media = notas.mean()
print(f"Media notas: {media:.2f}")


if notas.mean() >= 7:
    print(f"Turma com bom desempenho {media:.2f}")
else:
    print("Turma precisa melhorar")

print(notas)
```

```python
#2. Exercício 2 – Alunos aprovados Crie o array: [5.5, 7.0, 8.5, 6.0, 9.5, 4.5, 7.5] Percorra as notas e, para cada uma, utilize if/else para exibir se o aluno está:
#• Aprovado — nota maior ou igual a 7;
#• Reprovado — nota menor que 7.
#Exemplo: Nota: 5.5 - Reprovado     Nota: 7.0 - Aprovado  Nota: 8.5 - Aprovado

notas = np.array([5.5, 7.0, 8.5, 6.0, 9.5, 4.5, 7.5])

# Lista de notas dos alunos
notas = [5.5, 7.0, 8.5, 6.0, 9.5, 4.5, 7.5]

# O laço 'for' vai pegar uma nota por vez, isoladamente
for nota in notas:

    # Análise individual da nota atual do ciclo
    if nota >= 7.0:
        situacao = "Aprovado"
    else:
        situacao = "Reprovado"

    # Exibe o resultado da nota que está sendo analisada no momento
    print(f"Nota: {nota} - {situacao}")
```

```python
#3. Exercício 3 – Classificação das temperaturas Crie um array: [18, 25, 32, 15, 28, 35, 20]
#Percorra o array e classifique cada temperatura utilizando if/elif/else:
#• Menor que 20: Frio;     • Entre 20 e 29: Agradável;  • 30 ou mais: Quente.

temperaturas = np.array([18, 25, 32, 15, 28, 35, 20])

temperaturas = [18, 25, 32, 15, 28, 35, 20]

for temperatura in temperaturas:
  if temperatura < 20:
    clima = "Frio"
  elif temperatura >= 20 and temperatura <= 29:
    clima = "Agradável"
  else:
    clima = "Quente"
  print(f"Clima: {temperatura} -{clima} ")
```

```python
#4. Exercício 4 – Maior e menor salário Crie um array: [2500, 1800, 3200, 2100, 4500, 2800] Encontre o maior e o menor salário.
#Depois, utilize if para verificar se o maior salário é superior a R$ 4.000,00.
#Caso seja, exiba:
#Existe salário acima de R$ 4.000,00 Caso contrário: Não existe salário acima de R$ 4.000,00

salarios = np.array([2500, 1800, 3200, 2100, 4500, 2800])

salarios = [2500, 1800, 3200, 2100, 4500, 2800]
maior_salario = max(salarios)
menor_salario = min(salarios)

print(f"Maior salario: R$ {maior_salario:.2f}")
print(f"Menor salario: R$ {menor_salario:.2f}")

if maior_salario > 4000:
  print("Existe salário acima de R$ 4.000,00")
else:
  print("Não existe salario acima R$ 4.000,00")
```

```python
#5. Exercício 5 – Produtos em promoção Crie um array contendo os preços: [50, 120, 80, 200, 150, 30] Percorra os preços e
# aplique a seguinte regra utilizando if/else:
# • Produtos com preço maior ou igual a R$ 100 recebem 10% de desconto;
# • Produtos abaixo de R$ 100 permanecem com o mesmo preço.
# Exiba o preço original e o preço final de cada produto

precos = np.array ([50, 120, 80, 200, 150, 30])

for preco in precos:
  if preco >= 100:
    print(f"Preço original:{preco} - preço final com desconto: {preco*0.90}")
  else:
    print(f"Preço original:{preco} - preço sem desconto")
```
