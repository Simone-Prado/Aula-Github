```python
# Operadores lógicos
numero1 = 1
numero2 = 2

#exemplo if com operador lógico AND
if numero1 == 1 and numero2 == 2:
 print("O valores são iguais!")
```

```python
# Exemplo de if com operador lógico OR
if numero1 > 0  or numero2 < 0:
 print("O número 1 é maior que zero, mas o número 2 não é menor que zero")
```

```python
# exemplo de if com operador lógico de NEGAÇÃO - NOT
if not numero2 < 0:
  print("O número é maior do que zero")
```

```python
# Exemplo de if com else
valor1 = 3

if valor1 >=3:
  print("O valor 1 é maior ou igual á 3")
else:
  print("O valor 1 não é maior ou igual à 3")
```

```python
# Exemplo de if com eli e else
idade = 61

if idade > 18:
  print("Jovem")
elif idade >= 18 and idade <= 60:
  print("Adulto")
else:
  print("Idoso")
```

```python
# Exemplo de if abreviado (simplificado)
dia = 10

if dia == 10 : print("Hoje é dia 10") # apenas if
print("Hoje é dia 10")if dia == 10 else print("hoje não dia 10")
```

```python
# Exemplo match/case
opcao = int(input("Informe a opção: "))

match opcao:
  case 1:
    print("Falar com atendente")
  case 2:
    print("Dúvidas frequentes")
  case _:
    print ("Opção inválida")

```

```python
# Estruturas de Repetição
# exemplo simples de WHILE
contador = 0

while(contador < 10 ):
  print(contador)

# vamor forçar a parada
  if contador == 4:
   break

  contador += 1
```

```python
# exemplo de repetição com continue
# pular uma execução
contador = 0

while (contador <10 ):
  contador +=1

  # pular a interação
  if contador == 2:
    continue

  print(contador)
```

```python
# exemplo de qhile com else
contador = 0

while (contador < 10):
  print(contador)
  contador += 1
else:
  print("O contador é maior do que 10")
```
