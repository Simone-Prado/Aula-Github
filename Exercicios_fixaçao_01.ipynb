```python
#Exercicio 1 - apresentação pessoal
# Solicitando as informações do usuário
nome = input("Digite seu nome: ")
idade = input ("Digite sua idade: ")
cidade = input ("Digite a cidade onde você mora: ")

# Exibindo a mensagem final formatada (usando f-string)
print(f"Olá, {nome}! Você tem {idade} anos e mora em {cidade}.")
```

```python
# Exercicio 2 - Calculadora Simples
# Solicitando os dois números inteiros ao usuário
numero1 = int(input("Digite o primeiro numero inteiro"))
numero2 = int(input("Digite o segundo número inteiro: "))

# Realizando os cálculos matemáticos
soma = numero1 + numero2
subtracao = numero1 - numero2
multiplicacao = numero1 * numero2
divisao = numero1 / numero2

#Exibindo os resultados na tela
print( "\n --- Resultados ---")
print(f"Soma: ", {soma})
print(f"Subtração: ",{subtracao})
print(f"Multiplicação: ", {multiplicacao})
print(f"Divisão: ", divisao)
```
